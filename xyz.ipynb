```python
def is_prime(num):
    # This function checks if a given number 'num' is prime.
    
    if num < 2:
        # Numbers less than 2 are not prime by definition.
        return False
    
    for i in range(2, int(num**0.5) + 1):
        # We only need to check for divisors up to the square root of 'num'.
        # If 'num' has a divisor greater than its square root, it must also have one smaller.
        # int(num**0.5) ensures we get an integer, and + 1 is to include the square root itself in the range.
        
        if num % i == 0:
            # If 'num' is divisible by 'i' with no remainder, then 'num' is not prime.
            return False
            
    # If the loop completes without finding any divisors, the number is prime.
    return True

print("Prime numbers between 1 and 1000:")
# This line prints a descriptive header before listing the prime numbers.

for number in range(1, 1001):
    # This loop iterates through numbers from 1 to 1000 (inclusive).
    
    if is_prime(number):
        # For each number, it calls the 'is_prime' function to check if it's prime.
        
        print(number)
        # If the function returns True (meaning the number is prime), it prints the number.
```

````python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

print("Prime numbers between 1 and 1000:")
for number in range(1, 1001):
    if is_prime(number):
        print(number)
````
